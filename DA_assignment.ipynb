SOLUTION OF A


```python
import numpy as np
import matplotlib.pyplot as plt

# Function
def f(x):
    return 3*x**2 - 6*x + 8

# Derivative
def df(x):
    return 6*x - 6


# Gradient Descent
def gradient_descent(x0, eta, iterations):
    x = x0
    x_values = [x]
    f_values = [f(x)]

    for i in range(iterations):
        x = x - eta * df(x)

        x_values.append(x)
        f_values.append(f(x))

    return x_values, f_values


# Initial value
x0 = 0


# --------------------------------------------------
# A1. 5 Iterations for eta = 0.1 and eta = 0.01
# --------------------------------------------------

for eta in [0.1, 0.01]:

    x_values, f_values = gradient_descent(x0, eta, 5)

    print("\nLearning Rate =", eta)
    print("Iteration\t x\t\t f(x)")

    for i in range(6):
        print(i, "\t\t", round(x_values[i], 6),
              "\t", round(f_values[i], 6))


# --------------------------------------------------
# A2. Minimum Value
# --------------------------------------------------

# f'(x) = 6x - 6
# Therefore, minimum occurs at x = 1

minimum_x = 1
minimum_value = f(minimum_x)

print("\nMinimum of the function:")
print("x =", minimum_x)
print("Minimum value =", minimum_value)


# --------------------------------------------------
# A3. Gradient Descent for eta = 0.01, 0.5, 1.0
# --------------------------------------------------

learning_rates = [0.01, 0.5, 1.0]

for eta in learning_rates:

    x_values, f_values = gradient_descent(x0, eta, 5)

    print("\nLearning Rate =", eta)
    print("Iteration\t x\t\t f(x)")

    for i in range(6):
        print(i, "\t\t", round(x_values[i], 6),
              "\t", round(f_values[i], 6))


# --------------------------------------------------
# A4. Plot Iteration vs Function Value
# --------------------------------------------------

plt.figure(figsize=(8, 5))

for eta in learning_rates:

    x_values, f_values = gradient_descent(x0, eta, 10)

    plt.plot(
        range(len(f_values)),
        f_values,
        marker='o',
        label="eta = " + str(eta)
    )

plt.xlabel("Iteration")
plt.ylabel("Function Value f(x)")
plt.title("Gradient Descent")
plt.legend()
plt.grid(True)
plt.show()


# --------------------------------------------------
# A5. Finding the Fastest Converging Learning Rate
# --------------------------------------------------

tolerance = 0.001

for eta in [0.01, 0.1]:

    x_values, f_values = gradient_descent(x0, eta, 1000)

    for i, x in enumerate(x_values):

        if abs(x - minimum_x) < tolerance:

            print("\nLearning Rate:", eta)
            print("Converged at iteration:", i)
            print("x =", x)
            print("f(x) =", f(x))

            break
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# Part B
# Function: f(x) = x^4 - 3x^3 + 2
# --------------------------------------------------

def f(x):
    return x**4 - 3*x**3 + 2


# Derivative
def df(x):
    return 4*x**3 - 9*x**2


# --------------------------------------------------
# Gradient Descent
# --------------------------------------------------

def gradient_descent(x0, eta, iterations):

    x = x0

    x_values = [x]
    f_values = [f(x)]

    for i in range(iterations):

        gradient = df(x)

        x = x - eta * gradient

        x_values.append(x)
        f_values.append(f(x))

    return x_values, f_values


# --------------------------------------------------
# Parameters
# --------------------------------------------------

learning_rate = 0.01
iterations = 50

starting_points = [-1, 2.5]


# --------------------------------------------------
# Run Gradient Descent for both starting points
# --------------------------------------------------

for x0 in starting_points:

    x_values, f_values = gradient_descent(
        x0,
        learning_rate,
        iterations
    )

    print("\n===================================")
    print("Starting Point =", x0)
    print("Learning Rate =", learning_rate)
    print("===================================")

    print("Iteration\t x\t\t f(x)")

    for i in range(len(x_values)):

        print(
            i,
            "\t\t",
            round(x_values[i], 6),
            "\t",
            round(f_values[i], 6)
        )

    print("\nFinal x =", x_values[-1])
    print("Minimum function value =", f_values[-1])


# --------------------------------------------------
# Plot for x0 = -1
# --------------------------------------------------

x_values1, f_values1 = gradient_descent(
    -1,
    learning_rate,
    iterations
)

plt.figure(figsize=(8, 5))

plt.plot(
    range(len(f_values1)),
    f_values1,
    marker='o'
)

plt.xlabel("Iteration")
plt.ylabel("Function Value f(x)")
plt.title("Gradient Descent for x0 = -1")
plt.grid(True)

plt.show()


# --------------------------------------------------
# Plot for x0 = 2.5
# --------------------------------------------------

x_values2, f_values2 = gradient_descent(
    2.5,
    learning_rate,
    iterations
)

plt.figure(figsize=(8, 5))

plt.plot(
    range(len(f_values2)),
    f_values2,
    marker='o'
)

plt.xlabel("Iteration")
plt.ylabel("Function Value f(x)")
plt.title("Gradient Descent for x0 = 2.5")
plt.grid(True)

plt.show()


# --------------------------------------------------
# Compare both starting points
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    range(len(f_values1)),
    f_values1,
    marker='o',
    label="x0 = -1"
)

plt.plot(
    range(len(f_values2)),
    f_values2,
    marker='o',
    label="x0 = 2.5"
)

plt.xlabel("Iteration")
plt.ylabel("Function Value f(x)")
plt.title("Gradient Descent Comparison")
plt.legend()
plt.grid(True)

plt.show()


# --------------------------------------------------
# Analytical Critical Points
# --------------------------------------------------

# f'(x) = 4x^3 - 9x^2
#
# 4x^3 - 9x^2 = 0
#
# x^2(4x - 9) = 0
#
# Critical points:
# x = 0
# x = 9/4 = 2.25

print("\n===================================")
print("Analytical Results")
print("===================================")

print("Critical point 1: x =", 0)
print("f(0) =", f(0))

print("Critical point 2: x =", 9/4)
print("f(2.25) =", f(2.25))
```
