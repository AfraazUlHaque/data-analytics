```python
data = [
    [2.0, 60, 45, "Fail"],
    [2.5, 65, 50, "Fail"],
    [3.0, 68, 52, "Fail"],
    [3.5, 70, 55, "Fail"],
    [4.0, 72, 58, "Pass"],
    [5.0, 75, 65, "Pass"],
    [5.5, 80, 70, "Pass"],
    [6.0, 82, 75, "Pass"],
    [6.5, 85, 78, "Pass"],
    [7.0, 88, 82, "Pass"]
]

print("Dataset loaded successfully")
print("Total samples:", len(data))
```

```python
features = [
    "Study Hours",
    "Attendance (%)",
    "Assignment Score"
]

classes = []

for row in data:
    if row[3] not in classes:
        classes.append(row[3])

print("Features:", features)
print("Classes:", classes)
```

```python
def mean(values):
    total = 0

    for value in values:
        total += value

    return total / len(values)
```

```python
def variance(values):

    m = mean(values)

    total = 0

    for value in values:
        total += (value - m) ** 2

    return total / len(values)
```

```python
class_data = {}

for c in classes:

    class_data[c] = []

    for row in data:

        if row[3] == c:
            class_data[c].append(row)

for c in classes:
    print(c, ":", class_data[c])
```

```python
for c in classes:

    print("\nClass:", c)

    for i in range(3):

        values = []

        for row in class_data[c]:
            values.append(row[i])

        m = mean(values)
        v = variance(values)

        print(features[i])
        print("Mean =", m)
        print("Variance =", v)
```

```python
priors = {}

total = len(data)

for c in classes:

    count = len(class_data[c])

    priors[c] = count / total

    print("P(" + c + ") =", priors[c])
```

```python
import math

def gaussian(x, m, v):

    numerator = 1

    denominator = math.sqrt(2 * math.pi * v)

    exponent = math.exp(
        -((x - m) ** 2) / (2 * v)
    )

    return (numerator / denominator) * exponent
```

```python
test = [4.5, 74, 62]

scores = {}

for c in classes:

    score = priors[c]

    print("\nClass:", c)

    for i in range(3):

        values = []

        for row in class_data[c]:
            values.append(row[i])

        m = mean(values)
        v = variance(values)

        likelihood = gaussian(
            test[i],
            m,
            v
        )

        print(
            "P(" + features[i] + " | " + c + ") =",
            likelihood
        )

        score = score * likelihood

    scores[c] = score

    print("Naive Bayes Score =", score)
```

```python
predicted_class = None
highest_score = -1

for c in classes:

    if scores[c] > highest_score:

        highest_score = scores[c]
        predicted_class = c

print("\nPredicted Class:", predicted_class)
```
