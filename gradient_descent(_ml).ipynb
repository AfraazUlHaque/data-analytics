```python
def f(x):
  return 3*x*x - 6*x + 8
def grad(x):
  return 6*x -6

x = 0
lr = 0.1


for i in range(100):
  x_new = x - lr * grad(x)
  print("iterations", i+1, ": x = ", x, "f(x) = ", f(x))

  if x_new-x <= 0.0000001 :
    break

  x = x_new


print(x)
print(f(x))
```
