```python
# Test Data
test_data = [
    ["high", "good", "yes", "java"],
    ["low", "poor", "no", "cpp"],
    ["medium", "avg", "yes", "python"]
]
```

```python
# Attribute Names
attributes = ["cgpa", "communication", "internship", "language"]

# Possible Values
attribute_values = {
    "cgpa": ["high", "medium", "low"],
    "communication": ["good", "avg", "poor"],
    "internship": ["yes", "no"],
    "language": ["java", "cpp", "python"]
}

classes = ["yes", "no"]
```

```python
# Step 1: Calculate Prior Probabilities


class_count = {"yes": 0, "no": 0}

for row in train_data:
    class_count[row[-1]] += 1

total = len(train_data)

prior = {}
for c in classes:
    prior[c] = class_count[c] / total

print("Prior Probabilities")
for c in classes:
    print(c, "=", prior[c])

print()


# Step 2: Calculate Conditional Probabilities

cond_prob = {}

for c in classes:
    cond_prob[c] = {}

    for i in range(len(attributes)):
        attr = attributes[i]
        cond_prob[c][attr] = {}

        for value in attribute_values[attr]:
            count = 0

            for row in train_data:
                if row[-1] == c and row[i] == value:
                    count += 1


            probability = (count + 1) / (class_count[c] + len(attribute_values[attr]))

            cond_prob[c][attr][value] = probability


# Step 3: Prediction Function


def predict(sample):

    probability = {}

    for c in classes:

        prob = prior[c]

        for i in range(len(attributes)):
            attr = attributes[i]
            value = sample[i]
            prob *= cond_prob[c][attr][value]

        probability[c] = prob

    if probability["yes"] > probability["no"]:
        return "yes", probability
    else:
        return "no", probability

```

```python
# Step 4: Test Dataset

print("Predictions")


for sample in test_data:
    prediction, probs = predict(sample)

    print("Test Sample :", sample)
    print("P(yes) =", probs["yes"])
    print("P(no)  =", probs["no"])
    print("Predicted Class =", prediction)
    print()
```
